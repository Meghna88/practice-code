## Basic Euler Method — Malaria Case Growth

We use:

$$ \frac{dC}{dt}=rC $$

and Euler's equation:
$$ Cn+1​=Cn​+Δt(rCn​) $$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# -----------------------------
# Model parameters
# -----------------------------
C0 = 100       # Initial malaria cases
r = 0.20       # Growth rate per day
dt = 1         # Time step = 1 day
days = 30      # Simulation period
```

```python
# -----------------------------
# Create time array
# -----------------------------
time = np.arange(0, days + dt, dt)

# Create array to store cases
C = np.zeros(len(time))

# Initial condition
C[0] = C0

```

```python
# -----------------------------
# Euler Method
# -----------------------------
for n in range(len(time) - 1):

    # Calculate rate of change
    dCdt = r * C[n]

    # Euler equation
    C[n + 1] = C[n] + dt * dCdt

```

```python
# -----------------------------
# Display results
# -----------------------------
for n in range(len(time)):
    print(f"Day {time[n]:.0f}: {C[n]:.2f} cases")

```

```python
# -----------------------------
# Plot results
# -----------------------------
plt.figure(figsize=(8, 5))

plt.plot(time, C, marker='o')

plt.xlabel("Time (days)")
plt.ylabel("Malaria Cases")
plt.title("Malaria Case Growth Using Euler Method")

plt.grid(True)
plt.show()
```
